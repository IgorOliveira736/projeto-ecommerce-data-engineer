```python
df_orders = (spark.read
    .format("csv")
    .option("header", "true") 
    .option("inferSchema", "true")  
    .load("/Volumes/workspace/default/ecommerce_raw/olist_orders_dataset.csv")
)

df_customers = (
    spark.read
    .format("csv")
    .option("header", "true") 
    .option("inferSchema", "true")  
    .load("/Volumes/workspace/default/ecommerce_raw/olist_customers_dataset.csv")
)

df_items = (
    spark.read
    .format("csv")
    .option("header", "true") 
    .option("inferSchema", "true")  
    .load("/Volumes/workspace/default/ecommerce_raw/olist_order_items_dataset.csv")
)

df_orders.write.format("delta").mode("overwrite").saveAsTable("bronze_orders")
df_customers.write.format("delta").mode("overwrite").saveAsTable("bronze_customers")
df_items.write.format("delta").mode("overwrite").saveAsTable("bronze_order_items")

print("Sucesso! O Spark processou os CSVs e criou as tabelas Delta na Camada Bronze.")
```
